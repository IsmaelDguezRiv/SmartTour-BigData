## 🛠️ Diagnóstico Técnico y Limpieza Urgente

Este dataset destaca por tener una **corrupción de caracteres ("mojibake") acumulada doble**, provocada por guardar y abrir el archivo repetidamente con codificaciones incompatibles (como mezclar UTF-8, ANSI y Latin-1). 

Requiere una corrección inmediata de los nombres de sus columnas antes de trabajar con él:

### 1. 🔀 Reparación de Columnas Corruptas
A simple vista, tres de las seis columnas tienen nombres deformados por los acentos y los bytes invisibles de origen (BOM):
* `Ã¯Â»Â¿Provincia` ➔ Debería ser **`Provincia`**
* `Nombre del recurso turÃƒÂ­stico` ➔ Debería ser **`Nombre del recurso turístico`**

```python
#Eliminar columnas que están prácticamente vacías o no aportan valor
columnas_a_eliminar = ['Comarca', 'Mancomunidades', 'Entidades_Locales_Menores', 'CoordenadaX', 'CoordenadaY']
municipios_limpio = municipios.drop(columns=columnas_a_eliminar)
#Convertir coordenadas de texto (str) a decimales (float)
columnas_geo = ['Longitud', 'Latitud']
for col in columnas_geo:
    # Reemplazamos comas por puntos por si el formato regional fuera europeo
    municipios_limpio[col] = municipios_limpio[col].astype(str).str.replace(',', '.')
    municipios_limpio[col] = pd.to_numeric(municipios_limpio[col], errors='coerce')
#Formatear el Cod_INE como texto de 5 dígitos (evita que se borren ceros a la izquierda)
municipios_limpio['Cod_INE'] = municipios_limpio['Cod_INE'].astype(str).str.zfill(5)

print(municipios_limpio.info())
print(municipios_limpio.head(10))
```

```python
# 1. HACER UNA COPIA DE SEGURIDAD
hoteles_limpio = hoteles.copy()

# ==========================================
# PASO 1: CORREGIR EL ENCODING Y NOMBRAR COLUMNAS
# ==========================================
# Renombramos explícitamente las columnas rotas para corregir el "mojibake"
nuevos_nombres = {
    'ï»¿Establecimiento': 'Establecimiento',
    'CategorÃ­a': 'Categoría',
    'DirecciÃ³n': 'Dirección',
    'TelÃ©fono 1': 'Teléfono_1',
    'TelÃ©fono 2': 'Teléfono_2',
    'TelÃ©fono 3': 'Teléfono_3',
    'PosiciÃ³n': 'Posición',
    'C.Postal': 'Codigo_Postal'
}
hoteles_limpio.rename(columns=nuevos_nombres, inplace=True)

# Si hay textos rotos dentro de las filas (ej. "Categoría" con caracteres raros), los reparamos:
columnas_texto = ['Establecimiento', 'Tipo', 'Categoría', 'Nombre', 'Dirección', 'Provincia', 'Municipio', 'Localidad']
for col in columnas_texto:
    if col in hoteles_limpio.columns:
        hoteles_limpio[col] = hoteles_limpio[col].astype(str).str.encode('latin1', errors='ignore').str.decode('utf-8', errors='ignore')
        # Limpiamos espacios en blanco accidentales al principio o final
        hoteles_limpio[col] = hoteles_limpio[col].str.strip()

# ==========================================
# PASO 2: ELIMINAR COLUMNAS BASURA O VACÍAS
# ==========================================
# Eliminamos la columna 27 (que está vacía) y otras residuales si no las vas a usar
columnas_a_eliminar = ['Column 27', 'Q Calidad'] 
hoteles_limpio.drop(columns=[c for c in columnas_a_eliminar if c in hoteles_limpio.columns], inplace=True)

# ==========================================
# PASO 3: SANAR LOS TIPOS MIXTOS (CÓDIGO POSTAL Y TELÉFONOS)
# ==========================================
# Código Postal: Forzar a texto, quitar decimales ".0" si se le colaron, y rellenar con ceros (deben ser 5 dígitos)
if 'Codigo_Postal' in hoteles_limpio.columns:
    hoteles_limpio['Codigo_Postal'] = (
        hoteles_limpio['Codigo_Postal']
        .astype(str)
        .str.replace(r'\.0$', '', regex=True) # Elimina el ".0" si Pandas lo leyó como float
        .str.strip()
    )
    # Reemplazamos los 'nan' o vacíos por valores nulos reales de numpy
    hoteles_limpio['Codigo_Postal'] = hoteles_limpio['Codigo_Postal'].replace(['nan', 'NaN', ''], np.nan)
    # Rellenamos con ceros a la izquierda los códigos postales válidos (ej: "8001" -> "08001")
    hoteles_limpio['Codigo_Postal'] = hoteles_limpio['Codigo_Postal'].apply(lambda x: x.zfill(5) if pd.notnull(x) else x)

# Teléfonos: Limpiar y homogeneizar como texto
for t in ['Teléfono_1', 'Teléfono_2', 'Teléfono_3']:
    if t in hoteles_limpio.columns:
        hoteles_limpio[t] = hoteles_limpio[t].astype(str).str.replace(r'\.0$', '', regex=True).str.strip()
        hoteles_limpio[t] = hoteles_limpio[t].replace(['nan', 'NaN', ''], np.nan)

        # ==========================================
# PASO 4: VALIDAR COORDENADAS GEOGRÁFICAS
# ==========================================
# Aseguramos que Latitud y Longitud sean flotantes puros para tus mapas
for geo in ['GPS.Longitud', 'GPS.Latitud']:
    if geo in hoteles_limpio.columns:
        hoteles_limpio[geo] = pd.to_numeric(hoteles_limpio[geo], errors='coerce')

# ==========================================
# PASO 5: TRATAMIENTO DE CAMPOS VACÍOS (OPCIONAL)
# ==========================================
# Rellenar con "No Disponible" campos categóricos vacíos si prefieres no tener NaNs
hoteles_limpio['Categoría'] = hoteles_limpio['Categoría'].fillna('Sin Categoría')
hoteles_limpio['web'] = hoteles_limpio['web'].fillna('No Disponible')

# ==========================================
# VERIFICACIÓN FINAL
# ==========================================
print("--- Dataset de Hoteles Sanado Exitosamente ---")
print(hoteles_limpio.info())
```
