```python
from itertools import product

#camisetas = ["C1", "C2", "C3"]
#pantalones = ["P1", "P2"]
#zapatos = ["Z1", "Z2"]

camisetas = ["camiseta rick y morty", "camisa lino ", "polo zara"]
pantalones = ["vaqueros", "chandal"]
zapatos = ["zapatillas sport", "crocs"]

outfits = list(product(camisetas, pantalones, zapatos))

print ("Outfits posibles:")
for outfit in outfits:
    print(outfit)
print(f"Número total de outfits: {len(outfits)}")
```
