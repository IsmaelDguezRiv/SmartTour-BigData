```python
import csv 

def crear_csv(nombre_archivo, datos):
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        escritor_csv = csv.writer(archivo)
        escritor_csv.writerows(datos)

    print("Archivo CSV creado exitosamente " + nombre_archivo)

def escribir_csv(nombre_archivo, datos):
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        escritor_csv = csv.writer(archivo)
        #filas = [[alumno] for alumno in datos] organizacion en filas para que se vea en una columna 
        escritor_csv.writerows([datos]) #al recorrer las filas, es como lista para unificarlo
        
    print("Datos agregados al archivo CSV: " + nombre_archivo)

grupo1 = {"alumno1", "alumno2", "alumno3", "alumno4", "alumno7"}
grupo2 = {"alumno1", "alumno3", "alumno8", "alumno7", "alumno10"}


#cargamos los datos de un csv creamos 2 csv con 5 filas - alumnos

grupo3 = grupo1 | grupo2
grupo4 = grupo1 & grupo2
grupo5 = grupo1 - grupo2
grupo6 = grupo2 ^ grupo1

print("Grupo 3 (Unión):", grupo3)
print("Grupo 4 (Intersección):", grupo4)       
print("Grupo 5 (Diferencia):", grupo5)
print("Grupo 6 (Diferencia simétrica):", grupo6)

crear_csv("grupoUnion.csv", grupo3)
crear_csv("grupoInterseccion.csv", grupo4)
crear_csv("grupoDiferencia.csv", grupo5)
crear_csv("grupoDifSimetrica.csv", grupo6)

escribir_csv("grupoUnion.csv", grupo3)
escribir_csv("grupoInterseccion.csv", grupo4)
escribir_csv("grupoDiferencia.csv", grupo5)
escribir_csv("grupoDifSimetrica.csv", grupo6)

   
```
