```python
import pandas as pd

# Ventas (CSV)
clientes = pd.read_csv("../datasets/Lab03b/clientes.csv")

# Productos (Excel)
ventas = pd.read_excel("../datasets/Lab03b/ventas.xlsx")

# Opiniones (JSON)
contacto = pd.read_json("../datasets/Lab03b/contacto.json")

print(f"Los clientes antes de duplicados: {len(clientes)}")
clientes = clientes.drop_duplicates()
print(f"Los clientes despues de duplicados: {len(clientes)}")
print(ventas)
print(contacto)
```

```python
clientes["nombre"] = clientes["nombre"].str.strip().str.capitalize()
print(clientes)
print("---------------------------------------------")
clientes["ciudad"] = clientes["ciudad"].str.strip().str.title()
print(clientes)
```

```python
clientes["fecha_alta"] = pd.to_datetime(
    clientes["fecha_alta"], format = "%d/%m/%Y", errors = "coerce"
).dt.strftime("%Y-%m-%d")
print(clientes["fecha_alta"])
```

```python
contacto = contacto.rename(columns={"cliente_id":"id_cliente"})
print(contacto)
```

```python
integrado = clientes.merge(ventas, on="id_cliente", how="left")
integrado = integrado.merge(contacto, on="id_cliente", how="left")

print(integrado)
```

```python
integrado.to_csv("../datasets/Lab03b/datosIntegrados.csv")
```
